#AS Dataset is small about 1100 Rows and dropping more rows will make it even less and drop models accuracy. So what i will do is i will take the median of the value and fill the missing values with that.

```python
df['Salary'].median()
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
```

```python
df['Salary'].isnull().sum()
```

```python
df['Age'].median()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Age'].isnull().sum()
```

```python
print("Remaining missing values:\n", df.isnull().sum())
```

Filled Age with its median 30.0 because dropping 211 rows would lose 20% of data. Filled Salary with its median $85,547.87 to preserve realistic compensation averages without deleting records.

```python
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Salary'] < lower_bound) | (df['Salary'] > upper_bound)]
print(f"Lower Bound: ${lower_bound:,.2f}")
print(f"Upper Bound: ${upper_bound:,.2f}")
print("Number of outliers:", len(outliers))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(df['Salary'], kde=True)
```

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='Performance_Score')
plt.title("Employee Headcount by Performance Score")
plt.show()
```

```python
sns.boxplot(data=df, x='Performance_Score', y='Salary')
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Age', y='Salary', hue='Remote_Work')
plt.title("Age vs Salary (Colored by Remote Work)")
plt.show()
```

```python
df['Join_Year'] = pd.to_datetime(df['Join_Date']).dt.year

plt.figure(figsize=(8, 4))
sns.lineplot(data=df, x='Join_Year', y='Salary', marker='o')
plt.title("Salary Trend Over Hiring Years (2020 - 2024)")
plt.show()
```

```python
corr_matrix = df[['Age', 'Salary', 'Phone']].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap of Numeric Features")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

order = ['Poor', 'Average', 'Good', 'Excellent']

sns.boxplot(data=df, x='Performance_Score', y='Salary', order=order, palette='Blues')

plt.title('Salary Distribution by Performance Score')
plt.xlabel('Performance Score')
plt.ylabel('Salary ($)')

plt.show()
```
